### Datos ya con factor de expansión - FEX_C

El "factor de expansión" en los datos de encuestas como los que proporciona el DANE (Departamento Administrativo Nacional de Estadística de Colombia) es un valor que indica la cantidad de personas de la población general que cada individuo en la muestra representa. Si tienes datos que ya incluyen el factor de expansión, significa que los datos ya están ponderados para reflejar la población general a la que se refieren

```python
# Contar personas, hogares, viviendas y población estimada usando FEX_C

# Total de registros/personas en la muestra
num_personas_muestra = df.shape[0]

# Total de viviendas únicas (una por DIRECTORIO)
num_viviendas = df['DIRECTORIO'].nunique()

# Total de hogares únicos (por DIRECTORIO + SECUENCIA_ENCUESTA)
num_hogares = df[['DIRECTORIO', 'SECUENCIA_ENCUESTA']].drop_duplicates().shape[0]

# Estimación total de personas representadas usando el factor de expansión
total_poblacion_estimado = df['FEX_C'].sum()

# Personas por hogar (muestra)
personas_por_hogar = df.groupby(['DIRECTORIO', 'SECUENCIA_ENCUESTA']).size().reset_index(name='personas_en_muestra')

print("📊 Resumen de la Encuesta")
print(f"👥 Total de personas en la muestra: {num_personas_muestra}")
print(f"🏠 Total de hogares: {num_hogares}")
print(f"🏘️ Total de viviendas: {num_viviendas}")
print(f"🌍 Total estimado de personas en la población: {int(total_poblacion_estimado):,}")
print("\nEjemplo de personas por hogar (muestra):")
print(personas_por_hogar.head())

```

Según lo anterior, lo ideal sería utilizar datos que tengan menos del 30% de valores nulos en la encuesta. Agruparemos por hogar y tendremos en cuenta solamente a las cabezas de hogar para ver si mejora el estadístico. Así no tendríamos que cambiar de variable

```python
def cobertura_por_columna_por_hogar(df, columns_to_check):
    """
    Calcula para cada columna cuántos hogares tienen al menos un valor no nulo.
    Agrupa por hogar (DIRECTORIO + SECUENCIA_ENCUESTA).
    """
    # Agrupar por hogar
    grupo_hogar = df.groupby(['DIRECTORIO', 'SECUENCIA_ENCUESTA'])

    # Evaluar si hay al menos un valor no nulo por columna (por hogar)
    cobertura_hogar = grupo_hogar[columns_to_check].apply(lambda x: x.notnull().any()).reset_index()

    # Total de hogares únicos
    total_hogares = cobertura_hogar.shape[0]

    # Calcular cobertura por columna
    resumen = []
    for col in columns_to_check:
        hogares_validos = cobertura_hogar[col].sum()
        hogares_sin_dato = total_hogares - hogares_validos
        resumen.append({
            "Columna": col,
            "Total hogares": total_hogares,
            "Hogares con dato": hogares_validos,
            "Hogares sin dato": hogares_sin_dato,
            "% con dato": round(hogares_validos / total_hogares * 100, 2)
        })

    return pd.DataFrame(resumen).sort_values(by="% con dato", ascending=False).reset_index(drop=True)

# ✅ Uso:
columns_to_analyze = [
    'factor_expansion',
    'actividad_semana_pasada',
    'tiene_trabajo_no_trabajo',
    'trabajo_no_remunerado',
    'busco_trabajo_ult_4_sem',
    'desea_trabajar',
    'trabajo_2_sem_ult_ano',
    'busco_post_empleo',
    'tipo_ocupacion',
    'teletrabajo_12_meses'
]

df_cobertura_columnas_hogar = cobertura_por_columna_por_hogar(df, columns_to_analyze)
df_cobertura_columnas_hogar
```

Posiblemente obtengamos un porcentaje superior, cercano al 70%, de datos no nulos para las personas cabeza de hogar.

```python
df_persona_principal = df[(df['SECUENCIA_P'] == 1) & (df['ORDEN'] == 1)]

# Contar por hogar si hay dato no nulo en 'tipo_ocupacion'
conteo_tipo_ocupacion = (
    df_persona_principal
    .groupby(['DIRECTORIO', 'SECUENCIA_ENCUESTA'])['tipo_ocupacion']
    .apply(lambda x: x.notnull().any())
    .reset_index(name='tipo_ocupacion_presente')
)

# Calcular totales
total_hogares = conteo_tipo_ocupacion.shape[0]
hogares_con_dato = conteo_tipo_ocupacion['tipo_ocupacion_presente'].sum()
hogares_sin_dato = total_hogares - hogares_con_dato
porcentaje_con_dato = round(hogares_con_dato / total_hogares * 100, 2)

# Mostrar resultado
pd.DataFrame([{
    "Total hogares": total_hogares,
    "Hogares con tipo_ocupacion": hogares_con_dato,
    "Hogares sin dato": hogares_sin_dato,
    "% con dato": porcentaje_con_dato
}])
```

```python
def analizar_variables_por_hogar(df, columnas):
    """
    Analiza una lista de columnas a nivel de hogar (filtrando persona principal).
    Devuelve una tabla ordenada por % de hogares con datos en cada variable.
    """
    # Filtrar solo personas principales del hogar
    df_principal = df[(df['SECUENCIA_P'] == 1) & (df['ORDEN'] == 1)]

    resultados = []

    for columna in columnas:
        conteo = (
            df_principal
            .groupby(['DIRECTORIO', 'SECUENCIA_ENCUESTA'])[columna]
            .apply(lambda x: x.notnull().any())
            .reset_index(name='dato_presente')
        )

        total_hogares = conteo.shape[0]
        hogares_con_dato = conteo['dato_presente'].sum()
        hogares_sin_dato = total_hogares - hogares_con_dato
        porcentaje = round(hogares_con_dato / total_hogares * 100, 2)

        resultados.append({
            "Variable": columna,
            "Total hogares": total_hogares,
            "Hogares con dato": hogares_con_dato,
            "Hogares sin dato": hogares_sin_dato,
            "% con dato": porcentaje
        })

    # Convertir a DataFrame y ordenar por porcentaje descendente
    df_resultado = pd.DataFrame(resultados).sort_values(by="% con dato", ascending=False).reset_index(drop=True)

    return df_resultado


variables = [
    'factor_expansion',
    'actividad_semana_pasada',
    'tiene_trabajo_no_trabajo',
    'trabajo_no_remunerado',
    'busco_trabajo_ult_4_sem',
    'desea_trabajar',
    'trabajo_2_sem_ult_ano',
    'busco_post_empleo',
    'tipo_ocupacion',
    'teletrabajo_12_meses'
]

# Ejecutar la función para cada variable y concatenar resultados
df_resumen = analizar_variables_por_hogar(df, variables)
df_resumen

```

Volveremos a explorar si existen otras variables que nos ayuden con un porcentaje superior al 70% de datos no nulos.

```python
import pandas as pd

def evaluar_variables_por_hogar(df, umbral_cobertura=65):
    # Filtrar persona principal del hogar
    df_principal = df[(df["SECUENCIA_P"] == 1) & (df["ORDEN"] == 1)].copy()
    df_principal["hogar_id"] = df_principal["DIRECTORIO"].astype(str) + "_" + df_principal["SECUENCIA_ENCUESTA"].astype(str)

    # Definir columnas a excluir
    cols_excluir = ['DIRECTORIO', 'SECUENCIA_ENCUESTA', 'SECUENCIA_P', 'ORDEN', 'hogar_id']
    columnas_a_evaluar = [col for col in df_principal.columns if col not in cols_excluir]

    total_hogares = df_principal["hogar_id"].nunique()
    resultados = []

    for variable in columnas_a_evaluar:
        try:
            hogares_con_dato = df_principal[df_principal[variable].notnull()]["hogar_id"].nunique()
            cobertura = round(hogares_con_dato / total_hogares * 100, 2)

            if cobertura >= umbral_cobertura:
                resultados.append({
                    "Variable": variable,
                    "Cobertura (%)": cobertura,
                    "Hogares válidos": hogares_con_dato,
                    "Total hogares": total_hogares
                })
        except Exception as e:
            print(f"❌ Error procesando variable {variable}: {e}")

    return pd.DataFrame(resultados).sort_values(by="Cobertura (%)", ascending=False).reset_index(drop=True)


df1 = pd.read_csv("../data/ds/Fuerza_de_trabajo.CSV", sep=";")
# Evaluar todas las variables del DataFrame
df_resultado = evaluar_variables_por_hogar(df1)

# Ver resultados

df_resultado.head(100)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, OrdinalEncoder
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns

# Paso 1: Filtrar personas principales
df_principal = df[(df["SECUENCIA_P"] == 1) & (df["ORDEN"] == 1)].copy()
df_principal["hogar_id"] = df_principal["DIRECTORIO"].astype(str) + "_" + df_principal["SECUENCIA_ENCUESTA"].astype(str)

# Paso 2: Variables para clustering
variables = ['tipo_ocupacion', 'tamano_empresa', 'aporta_pension', 'horas_trab_usual']

# Paso 3: Filtrar datos válidos
df_clust = df_principal[variables].dropna().copy()

# Paso 4: Codificar variables categóricas
categorical_cols = ['tipo_ocupacion', 'tamano_empresa', 'aporta_pension']
encoder = OrdinalEncoder()
df_clust[categorical_cols] = encoder.fit_transform(df_clust[categorical_cols])

# Paso 5: Escalar los datos
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_clust)

# Paso 6: Aplicar KMeans
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df_clust['cluster_ocupacion'] = kmeans.fit_predict(X_scaled)

# Paso 7: Asignar clúster a DataFrame principal
df_principal['cluster_ocupacion'] = pd.NA
df_principal.loc[df_clust.index, 'cluster_ocupacion'] = df_clust['cluster_ocupacion']

# Paso 8: Visualización
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_clust,
    x='horas_trab_usual',
    y='tamano_empresa',
    hue='cluster_ocupacion',
    palette='Set2',
    alpha=0.7
)
plt.title('Clusters: Horas trabajadas vs Tamaño empresa (Personas principales)')
plt.xlabel('Horas trabajadas usualmente')
plt.ylabel('Tamaño empresa (codificado)')
plt.grid(True)
plt.tight_layout()
plt.show()

```
