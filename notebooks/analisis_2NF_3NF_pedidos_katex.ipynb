
# Análisis de 2NF y 3NF para la tabla `Pedidos`

## Estructura de la tabla

$$
\text{Pedidos}(\text{PedidoID}, \text{MesaID}, \text{FechaPedido}, \text{PlatoID}, \text{NombrePlato}, \text{PrecioPlato})
$$

## Dependencias funcionales dadas

1.  
$$
\text{PedidoID} \rightarrow \text{MesaID}, \text{FechaPedido}
$$

2.  
$$
\text{PlatoID} \rightarrow \text{NombrePlato}, \text{PrecioPlato}
$$

---

## Paso 1: Verificación de 1NF

La tabla está en 1NF si no tiene atributos multivaluados ni repetidos.  
**Suponemos que la tabla cumple con 1NF.**

---

## Paso 2: Verificación de 2NF

La **2NF** requiere que la tabla esté en 1NF y que **todos los atributos no primos dependan completamente de la clave primaria**.

### Clave primaria candidata

Dado que un pedido puede contener múltiples platos, se infiere que la clave primaria es:

$$
\boxed{(\text{PedidoID}, \text{PlatoID})}
$$

### Dependencias parciales

1.  
$$
\text{PedidoID} \rightarrow \text{MesaID}, \text{FechaPedido}
$$  
es una dependencia **parcial** → **viola 2NF**

2.  
$$
\text{PlatoID} \rightarrow \text{NombrePlato}, \text{PrecioPlato}
$$  
también es una dependencia **parcial** → **viola 2NF**

---

## Paso 3: Descomposición a 2NF

Dividimos la tabla en tres relaciones para eliminar dependencias parciales:

### R₁: Pedidos

$$
\text{Pedidos}(\text{PedidoID}, \text{MesaID}, \text{FechaPedido})
$$

### R₂: Platos

$$
\text{Platos}(\text{PlatoID}, \text{NombrePlato}, \text{PrecioPlato})
$$

### R₃: DetallesPedido

$$
\text{DetallesPedido}(\text{PedidoID}, \text{PlatoID})
$$

---

## Paso 4: Verificación de 3NF

La **3NF** requiere que para toda dependencia funcional \( X \rightarrow A \), se cumpla al menos una de las siguientes:

1. \( X \rightarrow A \) es trivial  
2. \( X \) es superclave  
3. \( A \) es atributo primo  

Todas las dependencias en las relaciones resultantes cumplen con 3NF.

---

## Conclusión

La tabla original **no está en 2NF ni 3NF**.  
Tras la descomposición en:

- \( \text{Pedidos}(\text{PedidoID}, \text{MesaID}, \text{FechaPedido}) \)
- \( \text{Platos}(\text{PlatoID}, \text{NombrePlato}, \text{PrecioPlato}) \)
- \( \text{DetallesPedido}(\text{PedidoID}, \text{PlatoID}) \)

se garantiza cumplimiento de **2NF y 3NF**.
