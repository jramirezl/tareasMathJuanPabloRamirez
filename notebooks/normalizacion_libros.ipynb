## I
Sistema de gestion de biblioteca que almacena libros, autores y las ciudades de los autores:
Consideramos la siguiente tabla:

**Libros(ISBN, Título, AutorID, NombreAutor, CiudadAutor)**

Contexto:
- Llave primaria: ISBN
- Reglas del negocio:
  - Un autor puede escribir múltiples libros.
  - Cada autor vive en una única ciudad.

### Analisis de dependencias funcionales (DF)

Las dependencias funcionales que se derivan del enunciado son:

- $\text{ISBN} \rightarrow \text{Título, AutorID, NombreAutor, CiudadAutor}$
- $\text{AutorID} \rightarrow \text{NombreAutor, CiudadAutor}$

### Análisis de normalización

### ¿Está en 1NF?
Sí. Todos los atributos son atómicos

### ¿Está en 2NF?
No. Hay una sola clave primaria (ISBN), pero tenemos dependencias parciales implícitas del tipo 
`AutorID → NombreAutor, CiudadAutor`. Sin embargo, AutorID no es parte de la clave primaria, por tanto no hay violación de 2NF.

### ¿Está en 3NF?
No, porque `CiudadAutor` depende transitivamente de `ISBN` a través de `AutorID`.


$\text{ISBN} \rightarrow \text{AutorID} \rightarrow \text{CiudadAutor}$

dependencia transitivsa que viola la 3FN.


### Reestructuración: 3NF y BCNF

Dividimos la tabla en dos:

### Tabla 1: Libros


**Libros(ISBN, Título, AutorID)**
- Llave primaria: ISBN


### Tabla 2: Autores


**Autores(AutorID, NombreAutor, CiudadAutor)**
- Llave primaria: AutorID


### Comprobación de 3NF

Ambas tablas:

- Tienen sus dependencias funcionales con claves candidatas.
- No hay dependencias transitivas.
- cumplen 3NF


### Comprobación de BCNF

- En **Libros**, la única dependencia es $\text{ISBN} \rightarrow \text{Título, AutorID}$, donde ISBN es clave. Cumple BCNF.
- En **Autores**, $\text{AutorID} \rightarrow \text{NombreAutor, CiudadAutor}$, y AutorID es clave. También cumple BCNF.



 ## II

Dada una tabla de empleados con la siguiente estructura y dependencias funcionales:

**Empleados(EmpID, Nombre, SupervisorID, NombreSupervisor, Departamento)**

Evaluación de BCNF y Descomposición de la Relación `Empleados`

Dada la relación:

- $\text{Empleados}(\text{EmpID}, \text{Nombre}, \text{SupervisorID}, \text{NombreSupervisor}, \text{Departamento})$


Y las dependencias funcionales:

1. - $\text{EmpID} \rightarrow \text{Nombre}, \text{SupervisorID}, \text{Departamento}$

2. - $\text{SupervisorID} \rightarrow \text{NombreSupervisor}, \text{Departamento}$



### Paso 1: Determinar la clave primaria

De la primera dependencia funcional:

- $\text{EmpID} \rightarrow \text{Nombre}, \text{SupervisorID}, \text{Departamento}$

Sabemos que `EmpID` determina todos los atributos excepto `NombreSupervisor`.

Entonces, la clave primaria es:

- $\boxed{\text{EmpID}}$


### Paso 2: Verificación de BCNF


1. - $\text{EmpID} \rightarrow \text{Nombre}, \text{SupervisorID}, \text{Departamento} $
$(\Rightarrow)$ `EmpID` es clave → **Cumple BCNF**

2. - $\text{SupervisorID} \rightarrow \text{NombreSupervisor}, \text{Departamento}$
$(\Rightarrow)$ `SupervisorID` **no es clave** → **No cumple BCNF**



### Paso 3: Descomposición en BCNF

Descomponemos la tabla en dos relaciones:

### R₁: Supervisores


- $\text{R}_1(\text{SupervisorID}, \text{NombreSupervisor}, \text{Departamento})$

Con dependencia:

- $\text{SupervisorID} \rightarrow \text{NombreSupervisor}, \text{Departamento}$

Clave: `SupervisorID` → **BCNF**



### R₂: Empleados

$ \text{R}_2(\text{EmpID}, \text{Nombre}, \text{SupervisorID})$

Con dependencia:

$ \text{EmpID} \rightarrow \text{Nombre}, \text{SupervisorID} $

Clave: `EmpID` → **BCNF**


# III 
Análisis de 2NF y 3NF para la tabla `Pedidos`

## Estructura de la tabla

**Pedidos(PedidoID, MesaID, FechaPedido, PlatoID, NombrePlato, PrecioPlato)**

## Dependencias funcionales dadas

1. $\text{PedidoID} \rightarrow \text{MesaID}, \text{FechaPedido} $

2. $\text{PlatoID} \rightarrow \text{NombrePlato}, \text{PrecioPlato} $


### Paso 1: Verificación de 1NF

La tabla está en 1NF si no tiene atributos multivaluados ni repetidos.  
Suponemos que la tabla cumple con 1NF.


### Paso 2: Verificación de 2NF

La 2NF requiere que la tabla esté en 1NF y que todos los atributos no primos dependan completamente de la clave primaria.

### Clave primaria candidata

Dado que un pedido puede contener múltiples platos, se infiere que la clave primaria es:

$ \boxed{(\text{PedidoID}, \text{PlatoID})}$

### Dependencias parciales

1.  $\text{PedidoID} \rightarrow \text{MesaID}, \text{FechaPedido}$  
es una dependencia parcial → viola 2NF

2.  $\text{PlatoID} \rightarrow \text{NombrePlato}, \text{PrecioPlato}$  
también es una dependencia parcial → viola 2NF


### Paso 3: Descomposición a 2NF

Para eliminar dependencias parciales se divide la tabla en tres:

### R₁: Pedidos

**Pedidos(PedidoID,MesaID,FechaPedido)**

### R₂: Platos

**Platos(PlatoID,NombrePlato,PrecioPlato)**

### R₃: DetallesPedido

**DetallesPedido(PedidoID,PlatoID)**

### Paso 4: Verificación de 3NF

1. Todas las tablas cumplen BCNF.  
2. Cada tabla tiene una clave primaria que determina todos los atributos.
3. No hay dependencias transitivas ni parciales


Todas las dependencias en las relaciones resultantes cumplen con 3NF.
