### Punto 2
#### Encuentre los nombres de todas las películas que fueron dirigidas por 'C. Nolan'.
$$
\pi_{\text{pnombre}}(\sigma_{\text{director="C.Nolan"}}(\text{peliculas}))
$$

### Punto 3
#### Encuentre los nombres de todas las películas que se estrenaron después del año 2010.
$$
\pi_{\text{pnombre}}(\sigma_{\text{año > 2010}}(\text{peliculas}))
$$

### Punto 4
#### Encuentre los nombres de todas las películas de la categoría 'SciFi' que tienen una calificación mayor a 8.5.
$$
\pi_{\text{pnombre}}(\sigma_{\text{calificacion > 8.5 } \land \text{ categoria="SciFi"}}(\text{peliculas}))
$$

### Punto 5
#### Encuentre los identificadores (id) de todos los actores que actuaron en la película con nombre 'Interstellar'.
$$

\pi_{\text{actuo\_en.id\_actor}} 
    \left(
        \sigma_{
            \text{actuo\_en.id\_pelicula} = \text{peliculas.id } \land 
            \text{ peliculas.pnombre} = \text{"Interstellar"}
            } 
                \left( 
                    \text{actuo\_en} \times \text{peliculas} 
                \right)
    \right)
$$

### Punto 6
#### Encuentre los nombres de todos los actores que actuaron en alguna película dirigida por 'C. Nolan'.
$$
\pi_{\text{actores.anombre}} 
    \left(
        \sigma_{
            \text{actores.id} = \text{actuo\_en.id\_actor } \land 
            \text{ actuo\_en.id\_pelicula} = \text{peliculas.id } \land 
            \text{ peliculas.pdirector} = "C. Nolan"
  } 
  \left( 
    \text{actuo\_en} \times \text{peliculas} \times \text{actores}
  \right)
    \right)
$$
